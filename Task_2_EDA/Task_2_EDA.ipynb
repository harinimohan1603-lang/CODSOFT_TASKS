```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/used_car_sales.csv')
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
print("duplicate rows:",df.duplicated().sum())
```

```python
numeric_df=df.select_dtypes(include=['int64','float64'])
numeric_df.corr()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(),annot=True,cmap='coolwarm')
plt.title("correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Manufacturer Name', order=df['Manufacturer Name'].value_counts().index)
plt.xticks(rotation=45)
plt.title("Number of Cars by Manufacturer")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Car Type', order=df['Car Type'].value_counts().index)
plt.title("Car Type Distribution")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Price-$'], bins=30, kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Price-$'])
plt.title("Outliers in Car Price")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df.groupby('Manufacturer Name')['Price-$'].mean().sort_values().plot(kind='bar')
plt.title("Average Car Price by Manufacturer")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
top_manufacturers = df['Manufacturer Name'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_manufacturers.plot(kind='bar')
plt.title("Top 10 Manufacturers")
plt.xlabel("Manufacturer")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Energy')
plt.title("Energy Type Distribution")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Sales Rating'], bins=10, kde=True)
plt.title("Sales Rating Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Margin-%'])
plt.title("Margin Percentage")
plt.show()
```
